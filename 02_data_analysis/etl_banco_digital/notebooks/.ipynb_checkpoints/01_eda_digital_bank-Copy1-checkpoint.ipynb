### Passo 3: Exploração inicial

- Volume de registros

```python
def completudeSchema(df):
    print('Qtd. registros:', df.count(), '| Quantidade de colunas: ', len(df.columns))
    df.printSchema()

def completudeVar(df, df_nome="DataFrame"):
    colunas = df.columns
    resumo = []

    for col in colunas:
        total = df.count()
        nulos = df.filter(F.col(col).isNull()).count()
        duplicados = total - df.select(col).distinct().count()
        unicos = df.select(col).distinct().count()

        resumo.append({
            "coluna": col,
            "total": total,
            "duplicados": duplicados,
            "nulos": nulos,
            "valores_unicos": unicos
        })

    resumo_pd = pd.DataFrame(resumo)
    resumo_pd = resumo_pd.sort_values("duplicados", ascending=False).reset_index(drop=True)
    
    print(f"Resumo do {df_nome}:")
    display(resumo_pd)
```

```python
bases = [agencias, clientes, pix, contas, transacoes]
nomes = ["agencias", "clientes", "pix", "contas", "transacoes de cartão"]
```

```python
for nome, df in zip(nomes, bases):
    print('Schema da base:', nome)
    completudeSchema(df)
```

```python
for nome, df in zip(nomes, bases):
    completudeVar(df)
    # completudeVar(df, "coluna_especifica)
```

#### Passo 4: Padronização

```python
df = df.withColumnRenamed('atual', 'renomeada')
```

```python
# from pyspark.sql.types import DoubleType # df = df.withColumn("valor", F.col("valor").cast(DoubleType()))
df = df.withColumn("renda", F.col("renda").cast("float"))
df = df.withColumn("data_nascimento_formatada", date_format("data_nascimento", "dd/MM/yyyy"))
```

```python
# data sem separador
df_sim_date = df_sim
colunas_data = ["dt_nasc", "dt_obito"]

for c in colunas_data:
    df_sim_date = df_sim_date.withColumn(f"{c}_format", to_date(F.col(c), "ddMMyyyy"))
```

```python
# Verifica se o mesmo ID tem dados diferentes nas outras colunas
duplicados_detalhes = df.groupBy("id_cliente").count().filter("count > 1")
df.join(duplicados_detalhes, "id_cliente", "inner").show()
```

```python
# Tratando os duplicados, mantendo o cadastro mais recente. Define a janela: agrupa por ID e ordena pela data atualização mais recente
janela_cliente = Window.partitionBy("id_cliente").orderBy(F.col("dt_atualizacao").desc())

# Mantém apenas o registro mais recente de cada ID
clientes_unicos = (df.withColumn("row_num", F.row_number().over(janela_cliente))
                   .filter("row_num = 1")
                   .drop("row_num"))
```

```python
# Validando tratamento de dupicidade dos dados
total = clientes_unicos.count()
unicos = clientes_unicos.select("id_cliente").distinct().count()
duplicados = total - unicos
print(f"Total de registros duplicados: {duplicados}")
```

```python
# Investigação: Clientes com renda nula estão transacionando valores altos?
clientes_unicos.filter(F.col("renda").isNull()).summary().show()
```

```python
# 1. Realização da agregação por cliente (id_conta)
df = df.groupBy("id_conta").agg(
    F.count("id_pix").alias("quantidade_pix"),
    F.sum("valor").alias("valor_total_pix"),
    F.avg("valor").alias("ticket_medio_pix"),
    F.max("valor").alias("maior_pix"),
    F.min("valor").alias("menor_pix")
)

df.show(3)
```

```python
df = df.withColumn(
    "perfil_financeiro",
    F.when((F.col("valor_total_pix") + F.col("valor_total_cartao")) <= 3000, "Baixa")
     .when((F.col("valor_total_pix") + F.col("valor_total_cartao")) <= 10000, "Media")
     .otherwise("Alta")
)

df.groupBy("perfil_financeiro").count().show()
```
