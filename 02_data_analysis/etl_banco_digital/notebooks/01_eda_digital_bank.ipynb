### Passo 1: Importações

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Case").getOrCreate()

from pyspark.sql import functions as F
import pandas as pd
```

### Passo 2: Leitura das bases

```python
PATH = '/home/tatiane/repositorios/data-projects/02_data_analysis/etl_banco_digital/bronze/'

agencias = spark.read.csv(f"{PATH}agencias.csv", header=True)
clientes = spark.read.csv(f"{PATH}clientes.csv", header=True)
pix = spark.read.csv(f"{PATH}pix.csv", header=True)
contas = spark.read.csv(f"{PATH}contas.csv", header=True)
transacoes = spark.read.csv(f"{PATH}transacoes_cartao.csv", header=True)
```

### Passo 3: Exploração inicial

- Volume de registros

```python
bases = [agencias, clientes, pix, contas, transacoes]
nomes = ["agencias", "clientes", "pix", "contas", "transacoes de cartão"]
for nome, df in zip(nomes, bases):
    print('Schema da base de: ', nome)
    print('Qtd. registros:', df.count(), '| Quantidade de colunas: ', len(df.columns))
    df.printSchema()
```

- Conteúdo

```python
for col in agencias.columns:
    agencias.groupBy(col).count().orderBy(col).show()
```

```python
for col in clientes.columns:
    clientes.groupBy(col).count().orderBy(col).show()
```

```python
for col in pix.columns:
    pix.groupBy(col).count().orderBy(col).show()
```

```python
for col in contas.columns:
    contas.groupBy(col).count().orderBy(col).show()
```

```python
for col in transacoes.columns:
    transacoes.groupBy(col).count().orderBy(col).show()
```

- Resumo completo por base

```python
bases = [agencias, clientes, pix, contas, transacoes]
nomes = ["agencias", "clientes", "pix", "contas", "transacoes de cartão"]
for nome, df in zip(nomes, bases):
    print('Resumo:', nome)
    colunas = df.columns
    resumo = []

    for col in colunas:
        total = df.count()
        nulos = df.filter(F.col(col).isNull()).count()
        duplicados = total - df.select(col).distinct().count()
        unicos = df.select(col).distinct().count()

        resumo.append({
            "coluna": col,
            "total": total,
            "duplicados": duplicados,
            "nulos": nulos,
            "valores_unicos": unicos
        })

    resumo_pd = pd.DataFrame(resumo)
    resumo_pd = resumo_pd.sort_values("duplicados", ascending=False).reset_index(drop=True)
    
    display(resumo_pd)
```
