# Target Engineering — Classificação de Valor Estratégico

## Objetivo

Criar uma variável alvo derivada (customer_value_class) que represente
o valor estratégico do cliente para o banco, com base em:

- Volume financeiro
- Intensidade de uso
- Relacionamento
- Engajamento

Essa abordagem permite classificar clientes em:
Low_Value, Medium_Value e High_Value.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
```

```python
# Carregamento

df = pd.read_csv("/home/tatiane/repositorios/data-projects/03_machine_learning/etl_customer_value_classfication/raw/data/BankChurners.csv")
```

```python
score_vars = [
    "CLIENTNUM",
    "Total_Trans_Amt",
    "Total_Trans_Ct",
    "Credit_Limit",
    "Months_on_book",
    "Months_Inactive_12_mon"
]

df_select = df[score_vars].copy()
df_select.head()
```

```python
id_col = df["CLIENTNUM"]

features = [
    "Total_Trans_Amt",
    "Total_Trans_Ct",
    "Credit_Limit",
    "Months_on_book",
    "Months_Inactive_12_mon"
]

df_features = df[features]
```

```python
# Normalização das variáveis
scaler = MinMaxScaler()

df_select = pd.DataFrame(
    scaler.fit_transform(df_select),
    columns=score_vars
)

df_select.head()
```

```python
df_select["customer_value_score"] = (
    0.4 * df_select["Total_Trans_Amt"] +
    0.25 * df_select["Total_Trans_Ct"] +
    0.20 * df_select["Credit_Limit"] +
    0.10 * df_select["Months_on_book"] -
    0.15 * df_select["Months_Inactive_12_mon"]
)

df_select["customer_value_score"].describe()
```

```python
# Criação de classes para classificação
df_select["customer_value_class"] = pd.qcut(
    df_select["customer_value_score"],
    q=3,
    labels=["Low_Value", "Medium_Value", "High_Value"]
)

df_select["customer_value_class"].value_counts()
```

#### Validação do Target

```python
# Distribuição das classes
df_select["customer_value_class"].value_counts(normalize=True)
```

```python
# Sanity check 1: Média por classe
df_select.groupby("customer_value_class")["Total_Trans_Amt"].mean()
```

```python
# Sanity check 2: Inatividade
df_select.groupby("customer_value_class")["Months_Inactive_12_mon"].mean()
```

```python
# Sanity check 3: Limite de crédito
df_select.groupby("customer_value_class")["Credit_Limit"].mean()
```

## Conclusão

O score derivado conseguiu segmentar os clientes em três grupos distintos, preservando coerência financeira e comportamental.

Clientes classificados como High_Value apresentam:
- Maior volume de transações
- Maior frequência de uso
- Maior limite de crédito
- Menor inatividade

A variável customer_value_class será utilizada como target no modelo de classificação multiclasse.
