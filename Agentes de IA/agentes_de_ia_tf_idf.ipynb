Rascunhos de aula sobre TF-IDF (Term Frequency Inverse Document Frequency), do curso 'Agentes de IA com python'

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# Lista de frases
frases = [
    "joão gosta de assistir files. Maria também gosta de filmes.",
    "João também gosta de assistir futebol.",
    "José foi assistir filmes com João e Maria."
]
```

```python
# Criação de vetor IT-IDF
Tfidf = TfidfVectorizer(norm=None)
```

```python
# Transformação das frases em matriz TF-IDF
Tfidf_x = Tfidf.fit_transform(frases).toarray()
print(f"Vetor de representação TF-IDT: \n{Tfidf_x}")
```

```python
# Criação do dataFrame com nomes das colunas
Tfidf_x = pd.DataFrame(data=Tfidf_x, columns=Tfidf.get_feature_names_out())
```

```python
# Visualização
Tfidf_x.head().round(2)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
from nltk.downloader import download
download('stopwords')
from nltk.corpus import stopwords
```

```python
#Definindo as stopwords
stop_words_portuguese = list(stopwords.words("portuguese"))
```

```python
#Instanciando o TfidfVectorizer com n-grams de 1 a N
Tfidf = TfidfVectorizer(ngram_range=(1,3),
stop_words=stop_words_portuguese,
strip_accents = 'unicode',
max_features=10,
norm='l1') # soma do vetor de cada frase vai ser 1 [melhor para semelhança de textos]; l2 é o 'padrão' para treinar ML
```

```python
# Ajustando e transformando as frases em uma matriz de contagem
Tfidf_X = Tfidf.fit_transform(frases).toarray()
```

```python
# Criando um DataFrame com os dados do Boku
Tfidf_X_df = pd.DataFrame(data=Tfidf_X, columns=Tfidf.get_feature_names_out())
```
